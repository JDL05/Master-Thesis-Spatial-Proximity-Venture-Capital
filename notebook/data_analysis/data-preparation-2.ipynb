All Investors from the same country

```python
# Create a new variable to check if all investors in the seed phase are from the same country
seed_help['seed_investors_all_same_country'] = seed_help.groupby('Company ID')['Country'].transform(lambda x: x.nunique() == 1)

# Map this variable to the main dataset
data['seed_investors_all_same_country'] = data['company_id'].map(seed_help.groupby('Company ID')['seed_investors_all_same_country'].first())
```

All Investors from the same country as investee

```python
# Create a new variable to check if all investors in the seed phase are from the same country as the startup
seed_help['seed_investors_all_same_country_as_startup'] = seed_help.apply(
    lambda row: row['Country'] == row['Headquarters Country'], axis=1
)

# Group by 'Company ID' and check if all investors for a company are from the same country as the startup
seed_help['seed_investors_all_same_country_as_startup'] = seed_help.groupby('Company ID')['seed_investors_all_same_country_as_startup'].transform(all)

# Map this variable to the main dataset
data['seed_investors_all_same_country_as_startup'] = data['company_id'].map(
    seed_help.groupby('Company ID')['seed_investors_all_same_country_as_startup'].first()
)
```

```python
# Create a new variable to check if all investors in the seed phase are from the same city
seed_help['seed_investors_all_same_city'] = seed_help.groupby('Company ID')['Seed Investor Same City'].transform(lambda x: x.all())

# Map this variable to the main dataset
data['seed_investors_all_same_city'] = data['company_id'].map(seed_help.groupby('Company ID')['seed_investors_all_same_city'].first())
```

```python
seed_help['Announced Date'] = pd.to_datetime(seed_help['Announced Date'])

seed_round_dates = seed_help.groupby('Company ID')['Announced Date'].agg(['min', 'max']).reset_index()
seed_round_dates.rename(columns={'min': 'earliest_seed_round_date', 'max': 'latest_seed_round_date'}, inplace=True)

data.rename(columns={'company_id': 'Company ID'}, inplace=True)

data = data.merge(seed_round_dates, on='Company ID', how='left')
```

European investors

```python
european_countries = [
    "Austria", "Belgium", "Bulgaria", "Cyprus", "Czech Republic", "Denmark", "Estonia", "Finland", "France", "Germany", "Greece", "Hungary","Iceland",
    "Ireland", "Italy", "Latvia", "Lithuania", "Liechtenstein", "Luxembourg", "Norway", "Poland", "Portugal", "Romania", "Slovenia",  "Slovakia (Slovak Republic)", "Spain",
    "Sweden", "Switzerland", "The Netherlands","Ukraine", "United Kingdom"
]

# Create the 'european_investor' variable: True if at least one investor is European
seed_help['european_investor'] = seed_help['Country'].isin(european_countries)

# Create the 'all_european_investor' variable: True if all investors are European
seed_help['all_european_investor'] = seed_help.groupby('Company ID')['Country'].transform(lambda x: x.isin(european_countries).all())

# Map these variables to the main dataset
data['european_investor'] = data['Company ID'].map(seed_help.groupby('Company ID')['european_investor'].any())
data['all_european_investor'] = data['Company ID'].map(seed_help.groupby('Company ID')['all_european_investor'].first())
```

Global Investor

```python
# Create the 'global_investor' variable: True if at least one investor is not European
seed_help['global_investor'] = ~seed_help['Country'].isin(european_countries)

# Create the 'all_global_investor' variable: True if all investors are not European
seed_help['all_global_investor'] = seed_help.groupby('Company ID')['Country'].transform(lambda x: ~x.isin(european_countries).all())

# Map these variables to the main dataset
data['global_investor'] = data['Company ID'].map(seed_help.groupby('Company ID')['global_investor'].any())
data['all_global_investor'] = data['Company ID'].map(seed_help.groupby('Company ID')['all_global_investor'].first())
```

US investor

```python
# Create the 'us_investor' variable: True if at least one investor is from the United States
seed_help['us_investor'] = seed_help['Country'] == 'United States'

# Create the 'all_us_investor' variable: True if all investors are from the United States
seed_help['all_us_investor'] = seed_help.groupby('Company ID')['Country'].transform(lambda x: (x == 'United States').all())

# Map these variables to the main dataset
data['us_investor'] = data['Company ID'].map(seed_help.groupby('Company ID')['us_investor'].any())
data['all_us_investor'] = data['Company ID'].map(seed_help.groupby('Company ID')['all_us_investor'].first())
```

DACH Investors

```python
dach_countries = [
    "Austria", "Germany", "Switzerland"
]

seed_help['dach_investor'] = seed_help['Country'].isin(dach_countries)

seed_help['all_dach_investor'] = seed_help.groupby('Company ID')['Country'].transform(lambda x: x.isin(dach_countries).all())

data['dach_investor'] = data['Company ID'].map(seed_help.groupby('Company ID')['dach_investor'].any())
data['all_dach_investor'] = data['Company ID'].map(seed_help.groupby('Company ID')['all_dach_investor'].first())
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# 1. Koordinaten anreichern
coords = data_global[['company_id', 'latitude', 'longitude']].drop_duplicates()
seed_help = seed_help.merge(coords, on='company_id', how='left')

# 2. Basis-Verbindungen
seed_help['co_conn']      = seed_help['num_investors'] - 1
seed_help['lead_conn']    = seed_help['co_conn'] * seed_help['Lead']
seed_help['co_only_conn'] = seed_help['co_conn'] - seed_help['lead_conn']

# 3. Haversine-Funktion (ASCII)
def hav(lat1, lon1, lat2, lon2):
    R = 6371  # Erdradius in km
    lat1_rad, lon1_rad, lat2_rad, lon2_rad = map(np.radians, (lat1, lon1, lat2, lon2))
    dlat = lat2_rad - lat1_rad
    dlon = lon2_rad - lon1_rad
    a = np.sin(dlat / 2)**2 + np.cos(lat1_rad) * np.cos(lat2_rad) * np.sin(dlon / 2)**2
    c = 2 * np.arcsin(np.sqrt(a))
    return R * c

# 4. Self-Join inkl. aller Vorgänger-Jahre + Filter auf ≤50 km
inv = seed_help[['Investor ID', 'company_id', 'Investment Year',
                 'co_only_conn', 'lead_conn', 'latitude', 'longitude']]

inv_other = inv.rename(columns={
    'company_id': 'other_company',
    'latitude':    'lat2',
    'longitude':   'lon2',
    'Investment Year': 'other_year'
})

# Merge nur auf Investor ID, dann nur Paare where other_year <= Investment Year
m = (
    inv
      .merge(inv_other, on='Investor ID')
      .query('company_id != other_company and other_year <= `Investment Year`')
)

# Distanz berechnen und auf ≤50 km filtern
m['dist'] = hav(m.latitude, m.longitude, m.lat2, m.lon2)
lm = m[m.dist <= 50]

# 5a) Lokales Portfolio (Anzahl unterschiedlicher Nachbar-Firmen)
portfolio = (
    lm.groupby('company_id')['other_company']
      .nunique()
      .rename('local_portfolio')
      .reset_index()
)

# 5b) Seed-Investoren pro Startup (für späteren Filter)
seed_inv_map = seed_help.groupby('company_id')['Investor ID'].unique().to_dict()

# 5c) Lokale Co-Investoren (ohne Seed-Investoren)
df_co = (
    lm[['company_id', 'other_company', 'Investment Year']]
      .merge(
          seed_help[['company_id', 'Investment Year', 'Investor ID', 'Lead']],
          left_on=['other_company', 'Investment Year'],
          right_on=['company_id', 'Investment Year'],
          suffixes=('', '_co')
      )
)
# Entferne ursprüngliche Seed-Investoren der Ursprungsfirma
df_co = df_co[
    ~df_co.apply(lambda r: r['Investor ID'] in seed_inv_map.get(r['company_id'], []), axis=1)
]

co_invs = (
    df_co.groupby('company_id')['Investor ID']
         .nunique()
         .rename('local_coinv')
         .reset_index()
)

lead_invs = (
    df_co[df_co['Lead'] == 1]
      .groupby('company_id')['Investor ID']
      .nunique()
      .rename('local_lead')
      .reset_index()
)

# 5d) Merge aller lokalen Metriken
lmets = (
    portfolio
      .merge(co_invs,  on='company_id', how='left')
      .merge(lead_invs, on='company_id', how='left')
      .fillna(0)
)

# 6. Local-Network-Score berechnen (Gewichtung [0.5, 0.5, 0])
weights = [0.5, 0.5, 0]
for col in ['local_portfolio', 'local_coinv', 'local_lead']:
    lmets[f'{col}_log'] = np.log1p(lmets[col])

scaled = MinMaxScaler().fit_transform(
    lmets[[f'{c}_log' for c in ['local_portfolio', 'local_coinv', 'local_lead']]]
)

for i, col in enumerate(['local_portfolio', 'local_coinv', 'local_lead']):
    lmets[f'{col}_s'] = scaled[:, i]

lmets['local_network_score'] = (
    lmets['local_portfolio_s'] * weights[0]
  + lmets['local_coinv_s']      * weights[1]
  + lmets['local_lead_s']       * weights[2]
)

# 7. Merge in data_global
data_global = data_global.merge(
    lmets[['company_id', 'local_portfolio', 'local_coinv', 'local_network_score']],
    on='company_id',
    how='left'
)

# Fehlende Werte auffüllen
for col in ['local_portfolio', 'local_coinv', 'local_network_score']:
    data_global[col] = data_global[col].fillna(0)

# 8. Binarisierung des Network-Scores
med  = data_global['local_network_score'].median()
q80  = data_global['local_network_score'].quantile(0.8)
data_global['high_local_net_median'] = (data_global['local_network_score'] > med).astype(int)
data_global['high_local_net_top20']  = (data_global['local_network_score'] >= q80).astype(int)

# 9. Neue Binaries für Portfolio und Co-Investoren
q80_port    = data_global['local_portfolio'].quantile(0.8)
q80_coinv   = data_global['local_coinv'].quantile(0.8)

data_global['high_portfolio_top20'] = (data_global['local_portfolio'] >= q80_port).astype(int)
data_global['high_coinv_top20']     = (data_global['local_coinv']    >= q80_coinv).astype(int)

# 10. Spalten-Reordering hinter 'max_seed_investor_pairwise_distance'
cols = data_global.columns.tolist()
idx  = cols.index('max_seed_investor_pairwise_distance') + 1

for col in [
    'local_portfolio',
    'local_coinv',
    'local_network_score',
    'high_local_net_median',
    'high_local_net_top20',
    'high_portfolio_top20',
    'high_coinv_top20'
]:
    cols.insert(idx, cols.pop(cols.index(col)))
    idx += 1

data_global = data_global[cols]
```

```python
# 1. Hilfs-DataFrame mit coords & score
coords = data_global[['company_id','latitude','longitude','local_network_score']].copy()
coords['key'] = 1

# 2. Cross-Join mit sich selbst, um Paare zu bilden
coords_other = coords.rename(columns={
    'company_id': 'other_id',
    'latitude':    'lat2',
    'longitude':   'lon2',
    'local_network_score': 'other_score'
})
coords_other['key'] = 1

pairs = coords.merge(coords_other, on='key').drop(columns='key')

# 3. Distanz berechnen
pairs['dist'] = hav(
    pairs.latitude, pairs.longitude,
    pairs.lat2,      pairs.lon2
)

# 4. Nur Nachbarn im 20 km-Radius (inkl. der Firma selbst, dist=0)
pairs20 = pairs[pairs.dist <= 20]

# 5. Für jede Firma den Score-Rang inkl. Eigen-Score bestimmen
#    Wir ziehen uns aus pairs20 pro company_id alle other_score (inkl. eigenem Score)
#    und ermitteln den 10. höchsten Wert als Schwellenwert
thresholds = (
    pairs20
      .groupby('company_id')['other_score']
      .apply(lambda scores: 
           sorted(scores, reverse=True)[9]     # 10. Wert, wenn ≥10 Scores
           if len(scores) >= 10 
           else sorted(scores, reverse=True)[-1]  # sonst kleinster Wert
      )
      .rename('peer_top10_threshold')
      .reset_index()
)

# 6. Schwellenwert in data_global mergen und Boolean erstellen
data_global = data_global.merge(thresholds, on='company_id', how='left')

data_global['in_peer_top10_20km'] = (
    data_global['local_network_score'] >= data_global['peer_top10_threshold']
).astype(int)

# 7. Optional: Für Firmen ohne Nachbarn (NaN) False setzen
data_global['in_peer_top10_20km'] = data_global['in_peer_top10_20km'].fillna(0).astype(int)

columns = data_global.columns.tolist()
high_coinv_top20_index = columns.index('high_coinv_top20')
columns.insert(high_coinv_top20_index + 1, columns.pop(columns.index('in_peer_top10_20km')))
data_global = data_global[columns]

data_global.drop(columns=['peer_top10_threshold'], inplace=True)
```

```python
# 1. Hilfs-DataFrame mit Koordinaten & local_coinv
coords_coinv = data_global[['company_id','latitude','longitude','local_coinv']].copy()
coords_coinv['key'] = 1

# 2. Cross-Join mit sich selbst
coords_coinv_other = coords_coinv.rename(columns={
    'company_id': 'other_id',
    'latitude':    'lat2',
    'longitude':   'lon2',
    'local_coinv': 'other_coinv'
})
coords_coinv_other['key'] = 1

pairs_coinv = coords_coinv.merge(coords_coinv_other, on='key').drop(columns='key')

# 3. Distanz berechnen
pairs_coinv['dist'] = hav(
    pairs_coinv.latitude,  pairs_coinv.longitude,
    pairs_coinv.lat2,      pairs_coinv.lon2
)

# 4. Filter nur Nachbarn im 20 km-Radius
pairs30_coinv = pairs_coinv[pairs_coinv.dist <= 30]

# 5. 80%-Perzentil des peer-local_coinv ermitteln
thresholds_coinv = (
    pairs30_coinv
      .groupby('company_id')['other_coinv']
      .quantile(0.8)
      .rename('peer_coinv_top20_threshold')
      .reset_index()
)

# 6. in_coinv_peer_top20_20km setzen
data_global = data_global.merge(thresholds_coinv, on='company_id', how='left')
data_global['in_coinv_peer_top20_30km'] = (
    data_global['local_coinv'] >= data_global['peer_coinv_top20_threshold']
).astype(int)

# 7. Firmen ohne genügend Nachbarn (NaN) als False markieren
data_global['in_coinv_peer_top20_30km'] = data_global['in_coinv_peer_top20_30km'].fillna(0).astype(int)

columns = data_global.columns.tolist()
in_peer_top10_20km_index = columns.index('in_peer_top10_20km')
columns.insert(in_peer_top10_20km_index + 1, columns.pop(columns.index('in_coinv_peer_top20_30km')))
data_global = data_global[columns]
```
